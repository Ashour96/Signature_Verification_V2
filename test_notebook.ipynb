```python
# # ============================================================
# # 1. Environment and imports
# # ============================================================
import matplotlib.pyplot as plt

from PIL import Image, ImageEnhance
import cv2

import torch
from torchvision import transforms

IMAGE_HEIGHT = 150
IMAGE_WIDTH = 220
```

```python
from src.model_load import get_model, device
from src.get_signature_embedding import get_signature_embedding
from src.squared_euclidean_distance import squared_euclidean_distance
from src.verify_signature import verify_signature
```

```python
model = get_model()
```

```python
questioned_path = r"F:\03 - Projects\8_Signature_Verification\Data\contract\Esraa_Org_1.png"
reference_path = r"F:\03 - Projects\8_Signature_Verification\Data\contract\Esraa_Org_2.png"

reference_img = Image.open(reference_path)
questioned_img = Image.open(questioned_path)



transform2 = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),
    transforms.Resize((IMAGE_HEIGHT, IMAGE_WIDTH))
])


reference_img = transform2(reference_img)
questioned_img = transform2(questioned_img)

reference_img = reference_img.convert("L")
questioned_img = questioned_img.convert("L")


# Make images slightly darker
brightness_factor = 0.015  # 1.0 = original, < 1.0 = darker

reference_img = ImageEnhance.Brightness(reference_img).enhance(brightness_factor)
questioned_img = ImageEnhance.Brightness(questioned_img).enhance(brightness_factor)

reference_img = reference_img.convert()

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(reference_img, cmap="gray")
plt.title("Reference Signature")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(questioned_img, cmap="gray")
plt.title("Questioned Signature")
plt.axis("off")

plt.tight_layout()
plt.show()


result = verify_signature(
    model=model,
    reference_path=reference_path,
    questioned_path=questioned_path,
    device=device
)

print("Decision:", result["decision"])
print("Distance:", result["distance"])
print("Threshold:", result["threshold"])
```
